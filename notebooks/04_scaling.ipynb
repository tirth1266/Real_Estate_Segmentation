```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
```

```python
# load client level table
df = pd.read_csv("../data/processed/client_features.csv",
                parse_dates = ["first_purchase","last_purchase"])
print(df.shape)
```

```python
# clustering features & check raw ranges
features = ["n_purchase","avg_price","avg_price_per_sqft","pct_Office"]
df[features].describe()
```

```python
# scale features to mean 0 & std 1
scaler = StandardScaler()
X_scaled = pd.DataFrame(scaler.fit_transform(df[features]), columns=features)
X_scaled.describe()
```

```python
import joblib
X_scaled.insert(0, "client_id",df["client_id"].values)
X_scaled.to_csv("../data/processed/clients_scaled.csv", index=False)
joblib.dump(scaler,"../data/processed/scaler.pkl")
print(X_scaled.shape)
```
